```python

# 1. Install library
!pip install -q mlxtend

# 2. Import libraries
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

# 3. Load dataset
df = pd.read_csv('/content/Disease_symptom_and_patient_profile_dataset.csv')

print("Dataset:")
print(df.head())
print("Columns:", df.columns.tolist())

# 4. Select symptom columns
symptoms = [
    'Fever', 'Cough', 'Fatigue',
    'Difficulty Breathing'
]

# Keep only columns available in the dataset
symptoms = [col for col in symptoms if col in df.columns]

# 5. Convert positive symptoms into transactions
transactions = []

for _, row in df.iterrows():
    items = [
        col for col in symptoms
        if str(row[col]).strip().lower() in ['yes', '1', 'true']
    ]
    if items:
        transactions.append(items)

# 6. One-hot encode transactions
te = TransactionEncoder()
encoded = te.fit(transactions).transform(transactions)
data = pd.DataFrame(encoded, columns=te.columns_)

# 7. Find frequent itemsets
frequent = apriori(
    data, min_support=0.05, use_colnames=True
)
print("\nFrequent Itemsets:")
print(frequent)

# 8. Generate association rules
if len(frequent) > 1:
    rules = association_rules(
        frequent, metric='confidence', min_threshold=0.3
    )

    print("\nAssociation Rules:")
    print(rules[['antecedents', 'consequents',
                 'support', 'confidence', 'lift']])
else:
    print("No frequent combinations found. Lower min_support.")
```
