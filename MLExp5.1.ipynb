```python
import pandas as pd
from collections import Counter


# ---------------- DATASET ----------------

data = pd.DataFrame({

    "Outlook": [
        "Sunny", "Sunny", "Overcast", "Rain",
        "Rain", "Rain", "Overcast", "Sunny",
        "Sunny", "Rain", "Sunny", "Overcast",
        "Overcast", "Rain"
    ],

    "Temperature": [
        "Hot", "Hot", "Hot", "Mild",
        "Cool", "Cool", "Cool", "Mild",
        "Cool", "Mild", "Mild", "Mild",
        "Hot", "Mild"
    ],

    "Humidity": [
        "High", "High", "High", "High",
        "Normal", "Normal", "Normal", "High",
        "Normal", "Normal", "Normal", "High",
        "Normal", "High"
    ],

    "Wind": [
        "Weak", "Strong", "Weak", "Weak",
        "Weak", "Strong", "Strong", "Weak",
        "Weak", "Weak", "Strong", "Strong",
        "Weak", "Strong"
    ],

    "PlayTennis": [
        "No", "No", "Yes", "Yes",
        "Yes", "No", "Yes", "No",
        "Yes", "Yes", "Yes", "Yes",
        "Yes", "No"
    ]
})


# ---------------- CONDITIONAL PROBABILITY ----------------

def conditional_probability(
    data, feature, value, target_value
):

    class_data = data[
        data["PlayTennis"] == target_value
    ]

    count_value = len(
        class_data[
            class_data[feature] == value
        ]
    )

    total_class = len(class_data)

    return count_value / total_class


# ---------------- NAÏVE BAYES ----------------

def predict(data, new_data):

    target = "PlayTennis"

    probabilities = {}

    for target_value in data[target].unique():

        # Prior probability
        probability = len(
            data[data[target] == target_value]
        ) / len(data)


        # Conditional probability
        for feature in new_data:

            value = new_data[feature]

            cond = conditional_probability(
                data,
                feature,
                value,
                target_value
            )

            probability = probability * cond


        probabilities[target_value] = probability


    # Highest probability = prediction
    prediction = max(
        probabilities,
        key=probabilities.get
    )

    return prediction, probabilities


# ---------------- USER INPUT ----------------

print("Enter New Weather Information")

outlook = input(
    "Outlook (Sunny/Overcast/Rain): "
).capitalize()

temperature = input(
    "Temperature (Hot/Mild/Cool): "
).capitalize()

humidity = input(
    "Humidity (High/Normal): "
).capitalize()

wind = input(
    "Wind (Weak/Strong): "
).capitalize()


# ---------------- NEW DATA ----------------

new_data = {

    "Outlook": outlook,
    "Temperature": temperature,
    "Humidity": humidity,
    "Wind": wind
}


# ---------------- PREDICTION ----------------

prediction, probabilities = predict(
    data,
    new_data
)


print("\nProbabilities:")
print(probabilities)

print(f"\nPrediction:{prediction} winning")
```

```python
import pandas as pd
import math

# Create dataset
data = pd.DataFrame({
    "Outlook": ["Sunny","Sunny","Overcast","Rain","Rain","Rain",
                "Overcast","Sunny","Sunny","Rain","Sunny","Overcast",
                "Overcast","Rain"],

    "Temperature": ["Hot","Hot","Hot","Mild","Cool","Cool",
                    "Cool","Mild","Cool","Mild","Mild","Mild",
                    "Hot","Mild"],

    "Humidity": ["High","High","High","High","Normal","Normal",
                 "Normal","High","Normal","Normal","Normal","High",
                 "Normal","High"],

    "Wind": ["Weak","Strong","Weak","Weak","Weak","Strong",
             "Strong","Weak","Weak","Weak","Strong","Strong",
             "Weak","Strong"],

    "PlayTennis": ["No","No","Yes","Yes","Yes","No",
                   "Yes","No","Yes","Yes","Yes","Yes",
                   "Yes","No"]
})


# Convert categorical values into numbers
data["Outlook"] = data["Outlook"].map({
    "Sunny": 0,
    "Overcast": 1,
    "Rain": 2
})

data["Temperature"] = data["Temperature"].map({
    "Hot": 0,
    "Mild": 1,
    "Cool": 2
})

data["Humidity"] = data["Humidity"].map({
    "High": 0,
    "Normal": 1
})

data["Wind"] = data["Wind"].map({
    "Weak": 0,
    "Strong": 1
})


# Take new input
outlook = input("Outlook (Sunny/Overcast/Rain): ").capitalize()
temperature = input("Temperature (Hot/Mild/Cool): ").capitalize()
humidity = input("Humidity (High/Normal): ").capitalize()
wind = input("Wind (Weak/Strong): ").capitalize()


# Convert new input into numbers
new_point = [
    {"Sunny": 0, "Overcast": 1, "Rain": 2}[outlook],
    {"Hot": 0, "Mild": 1, "Cool": 2}[temperature],
    {"High": 0, "Normal": 1}[humidity],
    {"Weak": 0, "Strong": 1}[wind]
]


# Choose K
k = 3

distances = []


# Calculate distance
for i in range(len(data)):

    point = [
        data.iloc[i]["Outlook"],
        data.iloc[i]["Temperature"],
        data.iloc[i]["Humidity"],
        data.iloc[i]["Wind"]
    ]

    distance = math.sqrt(
        (point[0] - new_point[0]) ** 2 +
        (point[1] - new_point[1]) ** 2 +
        (point[2] - new_point[2]) ** 2 +
        (point[3] - new_point[3]) ** 2
    )

    distances.append(
        (distance, data.iloc[i]["PlayTennis"])
    )


# Sort according to distance
distances.sort()


# Take K nearest neighbors
nearest = distances[:k]


# Count Yes and No
yes = 0
no = 0

for distance, label in nearest:

    if label == "Yes":
        yes += 1
    else:
        no += 1


# Prediction
if yes > no:
    prediction = "Yes"
else:
    prediction = "No"


print("\nNearest Neighbors:", nearest)
print("Yes =", yes)
print("No =", no)
print("Prediction =", prediction)
```
