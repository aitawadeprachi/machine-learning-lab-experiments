```python

import pandas as pd
import numpy as np



df = pd.read_csv("Crop_recommendation.csv")


print("First 7 Rows:")
print(df.head())


print("\nShape of Dataset:")
print(df.shape)


print("\nColumns:")
print(df.columns)

print("\nDataset Info:")
print(df.info())


print("\nSummary:")
print(df.describe())



numeric = df.select_dtypes(include=np.number)

print("\nMean:")
print(np.mean(numeric))

print("\nMaximum:")
print(np.max(numeric))

print("\nMinimum:")
print(np.min(numeric))

print("\nMultiple Columns:")
print(df[["N", "P"]])



print("\nTemperature > 25:")
print(df[df["temperature"] > 25])


print("\nSorted by Rainfall:")
print(df.sort_values(by="rainfall"))




print("Missing Values:")
print(df.isnull().sum())
df.dropna(inplace=True)
# Count duplicate rows
print("\nDuplicate Rows:")
print(df.duplicated().sum())
# Remove duplicate rows
df.drop_duplicates(inplace=True)
```
